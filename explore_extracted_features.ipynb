### Preprocess JSON like columns

```python
import json
import numpy as np
import pandas as pd

# Az eredeti df maradjon meg változatlanul.
df = df.copy()

def json_list(value):
    if isinstance(value, list):
        return value
    if pd.isna(value) or value == "":
        return []
    try:
        parsed = json.loads(value)
        return parsed if isinstance(parsed, list) else []
    except (TypeError, json.JSONDecodeError):
        return []

def numeric_values(items):
    values = []
    for item in items:
        if isinstance(item, dict):
            value = pd.to_numeric(item.get("value"), errors="coerce")
            if pd.notna(value):
                values.append(float(value))
    return np.asarray(values, dtype=float)

def entropy_features(items):
    values = numeric_values(items)

    if values.size == 0:
        return {
            "section_entropy_missing": 1,
            "section_entropy_count": np.nan,
            "section_entropy_mean": np.nan,
            "section_entropy_std": np.nan,
            "section_entropy_min": np.nan,
            "section_entropy_max": np.nan,
        }

    return {
        "section_entropy_missing": 0,
        "section_entropy_count": values.size,
        "section_entropy_mean": values.mean(),
        "section_entropy_std": values.std(ddof=0),
        "section_entropy_min": values.min(),
        "section_entropy_max": values.max(),
    }

MEM_EXECUTE = 0x20000000
MEM_READ = 0x40000000
MEM_WRITE = 0x80000000
RWX = MEM_EXECUTE | MEM_READ | MEM_WRITE

def section_characteristic_features(items):
    values = numeric_values(items).astype(np.uint32)

    if values.size == 0:
        return {
            "section_characteristics_missing": 1,
            "section_characteristics_count": np.nan,
            "executable_section_count": np.nan,
            "writable_section_count": np.nan,
            "rwx_section_count": np.nan,
        }

    return {
        "section_characteristics_missing": 0,
        "section_characteristics_count": values.size,
        "executable_section_count": int(((values & MEM_EXECUTE) != 0).sum()),
        "writable_section_count": int(((values & MEM_WRITE) != 0).sum()),
        "rwx_section_count": int(((values & RWX) == RWX).sum()),
    }

entropy_data = df["SectionEntropy"].map(json_list)
characteristics_data = df["SectionCharacteristics"].map(json_list)
import_data = df["ImportName"].map(json_list)

df = df.join(
    pd.DataFrame(
        entropy_data.map(entropy_features).tolist(),
        index=df.index,
    )
)

df = df.join(
    pd.DataFrame(
        characteristics_data.map(section_characteristic_features).tolist(),
        index=df.index,
    )
)

# Üres importlista csak akkor hiányzó, ha a jelentés mégis importokat jelez.
reported_functions = pd.to_numeric(
    df["NumberOfImportedFunctions"], errors="coerce"
)
reported_dlls = pd.to_numeric(df["ImportedDLLCount"], errors="coerce")

df["import_name_unique_count"] = import_data.map(lambda names: len(set(names)))
df["import_names_missing"] = (
    import_data.map(len).eq(0)
    & ((reported_functions > 0) | (reported_dlls > 0))
).astype("int8")

df.loc[df["import_names_missing"].eq(1), "import_name_unique_count"] = np.nan

# EntryPointSection: kategória, nem sorszám.
entry_section = df["EntryPointSection"].astype("string").str.strip()
df["entry_point_section_missing"] = (
    entry_section.isna() | entry_section.eq("")
).astype("int8")
df["entry_point_section_category"] = entry_section.fillna("__MISSING__")
df.loc[df["entry_point_section_category"].eq(""), "entry_point_section_category"] = "__MISSING__"

# Skew-elt méret- és darabszámváltozók logaritmikus változatai.
for column in [
    "FileSize",
    "SizeOfCode",
    "ImportedDLLCount",
    "NumberOfImportedFunctions",
    "import_name_unique_count",
]:
    df[column] = pd.to_numeric(df[column], errors="coerce")
    df[f"log1p_{column}"] = np.log1p(df[column].clip(lower=0))

df["size_of_code_ratio"] = (
    df["SizeOfCode"] / df["FileSize"]
).where(df["FileSize"] > 0)

df["size_of_code_gt_file_size"] = (
    df["SizeOfCode"].notna()
    & df["FileSize"].notna()
    & (df["SizeOfCode"] > df["FileSize"])
).astype("int8")

# A 6. sorhoz hasonló, hiányos jelentések jelölése.
df["report_complete_for_clustering"] = (
    df["section_entropy_missing"].eq(0)
    & df["section_characteristics_missing"].eq(0)
    & df["entry_point_section_missing"].eq(0)
).astype("int8")
```

### Clustering

```python
df_all = df.copy()  # teljes, eredeti adathalmaz megőrzése

df_cluster = df.loc[
    df["report_complete_for_clustering"].eq(1)
].copy()

print(f"Klaszterezéshez használt sorok: {len(df_cluster):,}")
print(f"Kizárt hiányos sorok: {len(df_all) - len(df_cluster):,}")
```

```python
# Csak teljes jelentések kerülnek az első klaszterezési körbe.
model_df = df.loc[df["report_complete_for_clustering"].eq(1)].copy()

# Azonosítók: megmaradnak, de nem lesznek X részei.
sample_ids = model_df[["SourceFile", "FileSHA256"]].copy()

# A 10 eredeti paraméter vagy az azokból képzett, numerikus változatok.
feature_columns = [
    # 1. FileSize
    "log1p_FileSize",

    # 2. FileEntropy
    "FileEntropy",

    # 3. NumberOfSections
    "NumberOfSections",

    # 4. SectionEntropy
    "section_entropy_count",
    "section_entropy_mean",
    "section_entropy_std",
    "section_entropy_max",

    # 5. SectionCharacteristics
    "section_characteristics_count",
    "executable_section_count",
    "writable_section_count",
    "rwx_section_count",

    # 6. SizeOfCode
    "log1p_SizeOfCode",
    "size_of_code_ratio",

    # 8. ImportedDLLCount
    "log1p_ImportedDLLCount",

    # 9. Number of imported functions
    "log1p_NumberOfImportedFunctions",

    # 10. ImportName
    "import_name_unique_count",
]

# 7. EntryPointSection: kategorikus változó -> one-hot, csak numerikus oszlopok.
entry_point_one_hot = pd.get_dummies(
    model_df["entry_point_section_category"],
    prefix="entry_point_section",
    dtype="int8",
)

X = pd.concat(
    [model_df[feature_columns], entry_point_one_hot],
    axis=1,
)

# Csak akkor távolítunk el további sort, ha valamely kiválasztott jellemző hiányzik.
missing_feature_rows = X.isna().any(axis=1)
X = X.loc[~missing_feature_rows].copy()
sample_ids = sample_ids.loc[X.index].copy()

# Ellenőrzés: minden klaszterezési bemenet numerikus.
assert X.select_dtypes(exclude="number").empty
assert not X.isna().any().any()

print(f"Klaszterezési minták száma: {len(X):,}")
print(f"Jellemzők száma: {X.shape[1]}")
```

```python
# scaling the numeric features using RobustScaler to reduce the impact of outliers
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import QuantileTransformer

# A ratio eredetije maradjon meg; új, korlátozott numerikus oszlop készül.
ratio_low, ratio_high = model_df["size_of_code_ratio"].quantile([0.01, 0.99])

model_df["size_of_code_ratio_capped"] = model_df[
    "size_of_code_ratio"
].clip(lower=ratio_low, upper=ratio_high)

numeric_columns = [
    "log1p_FileSize",
    "FileEntropy",
    "NumberOfSections",

    "section_entropy_mean",
    "section_entropy_std",
    "section_entropy_max",

    "executable_section_count",
    "writable_section_count",
    "rwx_section_count",

    "log1p_SizeOfCode",
    "size_of_code_ratio_capped",

    "log1p_ImportedDLLCount",
    "log1p_NumberOfImportedFunctions",
    "import_name_unique_count",
]

entry_point_columns = [
    column
    for column in model_df.columns
    if column.startswith("entry_point_section_")
]

X = model_df[numeric_columns + entry_point_columns].copy()

assert not X.isna().any().any()
# Convert the categorical entry-point section into numeric one-hot columns.
entry_point_one_hot = pd.get_dummies(
    model_df["entry_point_section_category"],
    prefix="entry_point_section",
    dtype="int8",
)

entry_point_columns = entry_point_one_hot.columns.tolist()

X = pd.concat(
    [
        model_df[numeric_columns],
        entry_point_one_hot,
    ],
    axis=1,
)

assert X.select_dtypes(exclude="number").empty
assert not X.isna().any().any()

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            QuantileTransformer(
                n_quantiles=min(1_000, len(X)),
                output_distribution="normal",
                random_state=42,
            ),
            numeric_columns,
        ),
        ("entry_point", "passthrough", entry_point_columns),
    ]
)

X_scaled = preprocessor.fit_transform(X)
```

### K-means

```python
from sklearn.cluster import KMeans

n_clusters = 5  # első próbafutás; később validáld több k értékkel

kmeans = KMeans(
    n_clusters=n_clusters,
    n_init=20,
    random_state=42,
)

cluster_labels = kmeans.fit_predict(X_scaled)

sample_ids = sample_ids.copy()
sample_ids["cluster"] = cluster_labels
```

### Validate cluster profiles quality

```python
from sklearn.metrics import (
    calinski_harabasz_score,
    davies_bouldin_score,
    silhouette_score,
)

# Use raw, interpretable values for profiling.
profile_columns = [
    "FileSize",
    "FileEntropy",
    "NumberOfSections",
    "section_entropy_mean",
    "section_entropy_std",
    "section_entropy_max",
    "executable_section_count",
    "writable_section_count",
    "rwx_section_count",
    "SizeOfCode",
    "size_of_code_ratio_capped",
    "ImportedDLLCount",
    "NumberOfImportedFunctions",
    "import_name_unique_count",
]

cluster_series = pd.Series(cluster_labels, index=X.index, name="cluster")
profile_df = model_df.loc[X.index, profile_columns].copy()
profile_df["cluster"] = cluster_series

cluster_sizes = cluster_series.value_counts().sort_index().rename("sample_count")
cluster_profile_median = profile_df.groupby("cluster")[profile_columns].median()
cluster_profile_mean = profile_df.groupby("cluster")[profile_columns].mean()

# Find the five features that most distinguish each cluster.
global_median = profile_df[profile_columns].median()
global_iqr = (
    profile_df[profile_columns].quantile(0.75)
    - profile_df[profile_columns].quantile(0.25)
).replace(0, np.nan)

robust_deviation = (cluster_profile_median - global_median) / global_iqr

top_drivers = []
for cluster, values in robust_deviation.iterrows():
    for feature in values.abs().nlargest(5).index:
        top_drivers.append(
            {
                "cluster": cluster,
                "feature": feature,
                "median": cluster_profile_median.loc[cluster, feature],
                "robust_deviation_iqr": values[feature],
            }
        )

top_drivers = pd.DataFrame(top_drivers).round(3)

# Entry-point section is categorical, so profile it separately.
entry_point_distribution = pd.crosstab(
    cluster_series,
    model_df.loc[X.index, "entry_point_section_category"],
    normalize="index",
)

dominant_entry_point = pd.DataFrame(
    {
        "most_common_entry_point": entry_point_distribution.idxmax(axis=1),
        "share": entry_point_distribution.max(axis=1),
    }
).round(3)

# Quality metrics. Silhouette uses a fixed 5,000-row sample for speed.
rng = np.random.default_rng(42)
sample_index = rng.choice(
    len(X_scaled),
    size=min(5_000, len(X_scaled)),
    replace=False,
)

quality_metrics = pd.Series(
    {
        "silhouette_score_sample": silhouette_score(
            X_scaled[sample_index],
            cluster_labels[sample_index],
        ),
        "davies_bouldin_score": davies_bouldin_score(X_scaled, cluster_labels),
        "calinski_harabasz_score": calinski_harabasz_score(
            X_scaled,
            cluster_labels,
        ),
    }
).round(4)

display(quality_metrics.to_frame("value"))
display(cluster_sizes.to_frame())
display(cluster_profile_median.round(3))
display(cluster_profile_mean.round(3))
display(top_drivers)
display(dominant_entry_point)
```
